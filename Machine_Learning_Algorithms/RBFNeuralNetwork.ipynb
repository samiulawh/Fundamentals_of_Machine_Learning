```python
import numpy as np
from sklearn.cluster import KMeans

# ----- Step 1: Create example training data -----
# Input: 2D points, Output: some nonlinear function
np.random.seed(0)
X = np.random.rand(20, 2)  # 20 samples, 2 features
y = np.sin(2 * np.pi * X[:, 0]) + np.cos(2 * np.pi * X[:, 1])  # target output

# ----- Step 2: Define number of RBF neurons -----
num_rbf = 3   # as you said
num_output = 1

# ----- Step 3: Find RBF centers using K-Means -----
kmeans = KMeans(n_clusters=num_rbf, random_state=0).fit(X)
centers = kmeans.cluster_centers_

# ----- Step 4: Compute widths (σ) -----
# We'll use the average distance between centers
dists = []
for i in range(num_rbf):
    for j in range(i+1, num_rbf):
        dists.append(np.linalg.norm(centers[i] - centers[j]))
sigma = np.mean(dists)
# (You could also make sigma_i individual if needed)

# ----- Step 5: Define the Gaussian RBF function -----
def rbf(x, c, s):
    return np.exp(-np.linalg.norm(x - c)**2 / (2 * s**2))

# ----- Step 6: Build the RBF layer output matrix H -----
H = np.zeros((X.shape[0], num_rbf))
for i in range(X.shape[0]):
    for j in range(num_rbf):
        H[i, j] = rbf(X[i], centers[j], sigma)

# ----- Step 7: Learn output weights using linear regression -----
# w = (H^T H)^(-1) H^T y
W = np.linalg.pinv(H).dot(y)

# ----- Step 8: Predict -----
def predict(x):
    h = np.array([rbf(x, centers[j], sigma) for j in range(num_rbf)])
    return h.dot(W)

# ----- Step 9: Test -----
test_input = np.array([0.2, 0.3])
pred = predict(test_input)
print("Predicted output:", pred)
```
