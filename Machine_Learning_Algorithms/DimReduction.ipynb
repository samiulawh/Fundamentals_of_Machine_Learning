```python
import numpy as np
import matplotlib.pyplot as plt
import math
```

```python
# Linear Discriminant Analysis
# data = np.array([(1,2,1),(3,2,6),(2,9,1),(8,1,3),(1,2,1),(2,1,3)])
data = np.array([(2,1),(2,6),(9,1),(1,3),(2,1),(1,3)])
print("Data: ", data)
full_mean = np.mean(data, axis=0)
# print("full_mean: ", full_mean)

n = len(data)
labels = np.array(["A" if i < n/2 else "B" for i in range(n)])

classes = np.unique(labels)  # find unique class names
# print("label: ", classes)
class_data = {}
for c in classes:
  # indx = np.where(labels == c) # gives output like (array([0, 1, 2, 3]),), we want like [0 1 2 3]
  indx = np.where(labels == c)[0]
  class_data[c] = data[indx,:]
  # print(f"class_data[{c}]")
  # print(class_data[c])
# print("class_data[A][i]", class_data['A'][0])

mean = {}
for c in classes:
  mean[c] = np.mean(class_data[c], axis=0) # if axis = 0, it calculate individual column mean, if 1 it rows like 2+1/2
  # print(f"mean[{c}]: ", mean[c])
# print("len(mean): ", len(mean['A']))

# within-class scatter
# summation(x - u)(x - u).T
def WithinClassScatter(a, m):
  Swo = np.zeros((a.shape[1], a.shape[1]))
  for i in range(len(a)):
    # print(f"a[{i}] - m: ", a[i], m)
    differ = (a[i] - m).reshape(-1,1)
    Swo += differ @ differ.T
  return Swo
Swo = {}
for c in classes:
    Swo[c] = WithinClassScatter(class_data[c], mean[c])
# total
Sw_total = np.zeros((data.shape[1], data.shape[1]))  # initialize 2x2
for c in classes:
    Sw_total += Swo[c]  # add each class scatter matrix
# print("Total within-class scatter Sw:\n", Sw_total)

#Between-Class Scatter Matrix
def BClass_Scatter(class_means, total_mean, class_sizes):
    Sb = np.zeros((total_mean.shape[0], total_mean.shape[0]))
    for c in class_means:
        diff = (class_means[c] - total_mean).reshape(-1,1)
        Sb += class_sizes[c] * (diff @ diff.T)
    return Sb
class_sizes = {c: len(class_data[c]) for c in classes}
Sb = BClass_Scatter(mean, full_mean, class_sizes)
# print("Between-class scatter Sb:\n", Sb)

LDA_matrix = np.linalg.inv(Sw_total) @ Sb
eigvals, eigvecs = np.linalg.eig(LDA_matrix)
# print("Eigenvalues:\n", eigvals)
# print("Eigenvectors (columns):\n", eigvecs)
w = eigvecs[:, 0].reshape(-1,1)
# Final
# Z = data * eigen
Z = data @ w
print(Z)
```

```python
# Principal component analysis
data2 = np.array([2,1,5,6,9,1,4])
data3 = np.array([3,1,4,9,1,1,2])

covariance_matrix = np.cov(data2, data3)
print("covariance_matrix", covariance_matrix)
#eigen
eigval, eigvec = np.linalg.eig(covariance_matrix)
print(eigval, eigvec)

# final
W = eigvec[:,0]
print("W: ", W)
d = np.column_stack((data2, data3))
Z = d @ W
# print("Projected data Z:", Z)
```
