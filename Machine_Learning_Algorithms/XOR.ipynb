```python
xor = {
    0: [(0, 0), (1, 1)],
    1: [(0, 1), (1, 0)]
}
weights = np.array([1,1,1,1])
bias = 0.1
def step(a):
  if a > 0:
    return 1
  return 0

print(len(xor.values())) # 2
def calcu(x):
  h = []
  for output, input in x.items():
    if output == 0:
      for pair in input:
        ho = pair[0] * weights[0] + pair[1] * weights[1] + bias
        st = step(ho)
        h.append(st)
    if output == 1:
      for pair in input:
        ho = pair[0] * weights[2] + pair[1] * weights[3] + bias
        st = step(ho)
        h.append(st)
  return h


print(calcu(xor))
```
