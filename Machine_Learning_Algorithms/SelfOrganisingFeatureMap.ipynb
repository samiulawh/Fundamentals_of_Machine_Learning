```python
# steps of SOM
# find euclidean distance
# choose neuron with min distance
# update weights, weights_new = weights_old + l_r*theta*(input - weights_old)
# theta = exp(-(pos_win - pos_i)**2/(2*sig**2))
# can update l_r, sigma > l_r = l_r*exp(-t/lembda), sig = sig*exp(-t/lembda), lembda = num_iterations/ln(sig)

input_vec = np.array([1, 2, 3])
weights = np.array([
    [1.1, 2.1, 0.7],
    [2.2, 2.1, 0.3],
    [0.1, 3.0, 0.4]
])

neuron0 = np.sqrt((input_vec[0] - weights[0][0])**2 + (input_vec[1] - weights[0][1])**2 + (input_vec[2] - weights[0][2])**2)
neuron1 = np.sqrt((input_vec[0] - weights[1][0])**2 + (input_vec[1] - weights[1][1])**2 + (input_vec[2] - weights[1][2])**2)
neuron2 = np.sqrt((input_vec[0] - weights[2][0])**2 + (input_vec[1] - weights[2][1])**2 + (input_vec[2] - weights[2][2])**2)
neurons = np.array([neuron0, neuron1, neuron2])

win_idx = np.argmin(neurons)
# print("winner_index: ", win_idx)
# weight update
l_r = 0.2
sig = 1
theta0 = np.exp( - (win_idx - 0)**2 / (2 * sig**2) )
print("theta0", theta0) # 1
theta1 = np.exp( - (win_idx - 1)**2 / (2 * sig**2) )
theta2 = np.exp( - (win_idx - 2)**2 / (2 * sig**2) )
theta = np.array([theta0, theta1, theta2])

weights[0] = weights[0] + l_r*theta[0]*(input_vec - weights[0])
weights[1] = weights[1] + l_r*theta[1]*(input_vec - weights[1])
weights[2] = weights[2] + l_r*theta[2]*(input_vec - weights[2])
# print("Updated Weights:", weights)

# Decay
# we are not doing epochs or iterations. but when we do, we can improve learning rate and sigma by using formula:
# lembda = total_iteration/np.log(sig)
# decay_l_r = l_r * np.exp( - t / lembda )
# decay_sig = sig * np.exp( - t / lembda )
```
