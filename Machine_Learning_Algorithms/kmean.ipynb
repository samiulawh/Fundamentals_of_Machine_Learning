```python
# kmean algorithm

# Data points
p1 = (185, 72)
p2 = (170, 56)
p3 = (168, 60)
p4 = (179, 68)
p5 = (182, 72)
p6 = (188, 77)

# Initial centroids (2D, NOT single numbers)
c1 = (185, 72)
c2 = (170, 56)

# Distance from centroid 1
d11 = sqrt((p1[0]-c1[0])**2 + (p1[1]-c1[1])**2)
d12 = sqrt((p2[0]-c1[0])**2 + (p2[1]-c1[1])**2)
d13 = sqrt((p3[0]-c1[0])**2 + (p3[1]-c1[1])**2)
d14 = sqrt((p4[0]-c1[0])**2 + (p4[1]-c1[1])**2)
d15 = sqrt((p5[0]-c1[0])**2 + (p5[1]-c1[1])**2)
d16 = sqrt((p6[0]-c1[0])**2 + (p6[1]-c1[1])**2)

# Distance from centroid 2
d21 = sqrt((p1[0]-c2[0])**2 + (p1[1]-c2[1])**2)
d22 = sqrt((p2[0]-c2[0])**2 + (p2[1]-c2[1])**2)
d23 = sqrt((p3[0]-c2[0])**2 + (p3[1]-c2[1])**2)
d24 = sqrt((p4[0]-c2[0])**2 + (p4[1]-c2[1])**2)
d25 = sqrt((p5[0]-c2[0])**2 + (p5[1]-c2[1])**2)
d26 = sqrt((p6[0]-c2[0])**2 + (p6[1]-c2[1])**2)

# Cluster assignment (manual, no loop)
cluster1 = []
cluster2 = []

cluster1.append(p1) if d11 < d21 else cluster2.append(p1)
cluster1.append(p2) if d12 < d22 else cluster2.append(p2)
cluster1.append(p3) if d13 < d23 else cluster2.append(p3)
cluster1.append(p4) if d14 < d24 else cluster2.append(p4)
cluster1.append(p5) if d15 < d25 else cluster2.append(p5)
cluster1.append(p6) if d16 < d26 else cluster2.append(p6)

print("Cluster 1:", cluster1)
print("Cluster 2:", cluster2)

cluster1 = [(185, 72), (179, 68), (182, 72), (188, 77)]
cluster2 = [(170, 56), (168, 60)]

# Calculate new centroids
new_c1_1 = (185+179+182+188)/4
new_c1_2 = (72+68+72+77)/4
new_c1 = (new_c1_1, new_c1_2)
new_c2_1 = (170+168)/2
new_c2_2 = (56+60)/2
new_c2 = (new_c2_1, new_c2_2)
print("new_c1:", new_c1)
print("new_c2:", new_c2)

# # Data points
# p1 = (185, 72)
# p2 = (170, 56)
# p3 = (168, 60)
# p4 = (179, 68)
# p5 = (182, 72)
# p6 = (188, 77)
points = [p1, p2, p3, p4, p5, p6]

bd11 = sqrt((185 - new_c1[0])**2 + (72 - new_c1[1])**2)
bd12 = sqrt((170 - new_c1[0])**2 + (56 - new_c1[1])**2)
bd13 = sqrt((168 - new_c1[0])**2 + (60 - new_c1[1])**2)
bd14 = sqrt((179 - new_c1[0])**2 + (68 - new_c1[1])**2)
bd15 = sqrt((182 - new_c1[0])**2 + (72 - new_c1[1])**2)
bd16 = sqrt((188 - new_c1[0])**2 + (77 - new_c1[1])**2)
updated_dist1 = [bd11, bd12, bd13, bd14, bd15, bd16]
# print("Distance from new centroid 1:", updated_dist1)

bd21 = sqrt((185 - new_c2[0])**2 + (72 - new_c2[1])**2)
bd22 = sqrt((170 - new_c2[0])**2 +(56 - new_c2[1])**2)
bd23 = sqrt((168 - new_c2[0])**2 + (60 - new_c2[1])**2)
bd24 = sqrt((179 - new_c2[0])**2 + (68 - new_c2[1])**2)
bd25 = sqrt((182 - new_c2[0])**2 + (72 - new_c2[1])**2)
bd26 = sqrt((188 - new_c2[0])**2 + (77 - new_c2[1])**2)
updated_dist2 = [bd21, bd22, bd23, bd24, bd25, bd26]
# print("Distance from new centroid 2:", updated_dist2)

updated_cluster1 = []
updated_cluster2 = []
for i in range(6):
  if updated_dist1[i] < updated_dist2[i]:
    updated_cluster1.append(points[i])
  else:
    updated_cluster2.append(points[i])
print("Updated cluster 1:", updated_cluster1)
print("Updated cluster 2:", updated_cluster2)
```

```python
# kmeans neural network, online kmeans
# 3 inputs neurons, 4 outputs neurons
# find h0 h1 h2 h3 > get winner (max) > update its weights > normalization > repeat
input = np.array([4,2,1])
w0,w1,w2,w3,w4,w5,w6,w7,w8,w9,w10,w11 = 0.6,0.4,0.3,0.1,0.1,0.1,0.2,0.2,0.2,0.4,0.4,0.2

print("1st iteration")
h0 = w0*input[0] + w1*input[1] + w2*input[2]
h1 = w3*input[0] + w4*input[1] + w5*input[2]
h2 = w6*input[0] + w7*input[1] + w8*input[2]
h3 = w9*input[0] + w10*input[1] + w11*input[2]
print("h0: ",h0,"h1: ",h1,"h2: ",h2,"h3: ",h3)
activation = np.array([h0,h1,h2,h3])
winner1 = max(activation)
print("winner1: ",winner1)

# weight update :
learning_rate = 0.1
w_old1 = np.array([w0,w1,w2])
w_new1 = [0,0,0]
w_new1[0] = w_old1[0] + learning_rate*(input[0] - w_old1[0])
w_new1[1] = w_old1[1] + learning_rate*(input[1] - w_old1[1])
w_new1[2] = w_old1[2] + learning_rate*(input[2] - w_old1[2])
print("w_new1: ",w_new1)
w0,w1,w2 = w_new1
print("w0: ",w0,"w1: ",w1,"w2: ",w2)

print("2nd iteration")
h0 = w0*input[0] + w1*input[1] + w2*input[2]
h1 = w3*input[0] + w4*input[1] + w5*input[2]
h2 = w6*input[0] + w7*input[1] + w8*input[2]
h3 = w9*input[0] + w10*input[1] + w11*input[2]
print("h0: ",h0,"h1: ",h1,"h2: ",h2,"h3: ",h3)
activation = np.array([h0,h1,h2,h3])
winner2 = max(activation)
print("winner2: ",winner2)

w_old2 = np.array([w0,w1,w2])
w_new2 = [0,0,0]
w_new2[0] = w_old2[0] + learning_rate*(input[0] - w_old2[0])
w_new2[1] = w_old2[1] + learning_rate*(input[1] - w_old2[1])
w_new2[2] = w_old2[2] + learning_rate*(input[2] - w_old2[2])
print("w_new2: ",w_new2)
w0,w1,w2 = w_new2
print("w0: ",w0,"w1: ",w1,"w2: ",w2)
```
